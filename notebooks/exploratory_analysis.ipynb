# Exploratory Analysis — Predictive Maintenance
Run `python main.py` first so `outputs/processed_data/cleaned_dataset.csv` exists.

```python
import sys, os
sys.path.append(os.path.abspath('..'))
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
sns.set_theme(style='whitegrid')
```

```python
df = pd.read_csv('../outputs/processed_data/cleaned_dataset.csv')
print(df.shape)
df.head()
```

```python
df.describe().T
```

```python
print('Failure rate:', df['Machine failure'].mean().round(4))
print(df['Machine failure'].value_counts())
```

```python
fig, ax = plt.subplots(figsize=(6,4))
df['Machine failure'].value_counts().plot(kind='bar', ax=ax, color=['#2e86ab','#d64541'])
ax.set_xticklabels(['Normal','Failure'], rotation=0); ax.set_ylabel('count'); ax.set_title('Machine failure distribution')
plt.show()
```

```python
fig, ax = plt.subplots(figsize=(12,9))
sns.heatmap(df.select_dtypes('number').corr(), annot=True, fmt='.2f', cmap='coolwarm', center=0, ax=ax)
plt.show()
```

```python
fig, axes = plt.subplots(1, 2, figsize=(13,5))
sns.boxplot(x='Machine failure', y='Torque [Nm]', data=df, ax=axes[0])
axes[0].set_title('Torque by failure')
sns.boxplot(x='Machine failure', y='Tool wear [min]', data=df, ax=axes[1])
axes[1].set_title('Tool wear by failure')
plt.show()
```

```python
fig, ax = plt.subplots(figsize=(9,5))
sns.scatterplot(x='Torque [Nm]', y='Rotational speed [rpm]', hue='Machine failure', data=df.sample(1500, random_state=42), alpha=.6, ax=ax)
plt.show()
```
